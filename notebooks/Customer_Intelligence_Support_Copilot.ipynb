import os

# Project directory
project_path = "/content/swiggy-instamart-customer-copilot"

# Create project folders
folders = [
    "data",
    "notebooks",
    "src",
    "policies",
    "outputs",
    "models"
]

for folder in folders:
    os.makedirs(
        os.path.join(project_path, folder),
        exist_ok=True
    )

print("Project folders created successfully!")

```python
files = [
    "generate_data.py",
    "mock_llm.py",
    "run_project.py",
    "Requirements.txt",
    "README.md",
    ".gitignore",
    "src/data_cleaning.py",
    "src/database.py",
    "src/segmentation.py",
    "src/churn_prediction.py",
    "src/triage.py",
    "src/rag_pipeline.py",
    "notebooks/Customer_Intelligence_Support_Copilot.ipynb",
    "policies/refund_policy.txt",
    "policies/delivery_policy.txt",
    "policies/cancellation_policy.txt"
]

for file in files:
    path = os.path.join(project_path, file)

    if not os.path.exists(path):
        open(path, "w").close()

print("All project files created!")
```

```python
for root, dirs, files in os.walk(project_path):
    level = root.replace(project_path, "").count(os.sep)

    print("  " * level + os.path.basename(root) + "/")

    for file in files:
        print("  " * (level + 1) + file)
```
